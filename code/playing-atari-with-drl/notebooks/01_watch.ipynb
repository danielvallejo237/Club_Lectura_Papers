## Baseline grids (committed in `assets/`)


```python
from pathlib import Path

GRID = {
    "cartpole": ("assets/cartpole_untrained_grid.gif", "assets/cartpole_trained_grid.gif"),
    "lunarlander": ("assets/lunarlander_untrained_grid.gif", "assets/lunarlander_trained_grid.gif"),
}[GAME]

for label, path in [("untrained grid", GRID[0]), ("trained grid", GRID[1])]:
    p = Path(path)
    print(label, p, "exists" if p.exists() else "MISSING — regenerate via README")
```

## Display grids


```python
from IPython.display import Image, display
from pathlib import Path

for label, path in [("untrained", GRID[0]), ("trained", GRID[1])]:
    p = Path(path)
    if p.exists():
        print(label)
        display(Image(filename=str(p)))
```
