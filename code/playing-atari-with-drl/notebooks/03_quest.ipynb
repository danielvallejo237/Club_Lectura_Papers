## Example: faster epsilon decay

Creates a copy of the baseline config with `epsilon_decay_steps: 10000`. Change `--config` to your quest config after editing `dqn/`.

```python
import yaml
from pathlib import Path

cfg = yaml.safe_load(open("configs/cartpole_paperlike.yaml"))
cfg["rl"]["epsilon_decay_steps"] = 10000
cfg["logging"]["output_dir"] = "outputs/quest_epsilon_fast"
out = Path("configs/quest_epsilon_fast.yaml")
yaml.dump(cfg, open(out, "w"), default_flow_style=False)
print("written", out)
```

```python
!python train.py --config configs/quest_epsilon_fast.yaml --seed 42 --overwrite
```
